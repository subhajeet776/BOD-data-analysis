```python
# Define features and target for regression
# Define features (X) and target (y)

X_reg = df[measurement_cols].drop('B.O.D. (mg/l)', axis=1)
y_reg = df['B.O.D. (mg/l)']
print('Regression features:', X_reg.columns.tolist())
print(f'\nFeature matrix shape: {X_reg.shape}')
print(f'Target series shape: {y_reg.shape}')
print('\nMissing values in features:')
print(X_reg.isnull().sum())
```

```python
# Define train-test splits
splits = [
    ('80:20', 0.20),
    ('70:30', 0.30),
    ('65:35', 0.35)
]

reg_results = []
reg_predictions = {}

for split_name, test_size in splits:

    # Split features and target
    X_train, X_test, y_train, y_test = train_test_split(
        X_reg, y_reg, test_size=test_size, random_state=42
    )

    # Fill missing values using training data medians
    medians = X_train.median()

    X_train = X_train.fillna(medians)
    X_test = X_test.fillna(medians)

    # Define models
    reg_models = {
        'Linear Regression': LinearRegression(),
        'Decision Tree': DecisionTreeRegressor(random_state=42),
        'Random Forest': RandomForestRegressor(
            n_estimators=100, random_state=42
        )
    }

    # Train and evaluate every model
    for model_name, model in reg_models.items():

        model.fit(X_train, y_train)
        y_pred = model.predict(X_test)

        mae = mean_absolute_error(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)
        rmse = np.sqrt(mse)
        r2 = r2_score(y_test, y_pred)

        reg_results.append({
            'Model': model_name,
            'Split': split_name,
            'MAE': round(mae, 4),
            'MSE': round(mse, 4),
            'RMSE': round(rmse, 4),
            'R2': round(r2, 4)
        })

        # Save predictions for the 80:20 visualisations
        if split_name == '80:20':
            reg_predictions[model_name] = (y_test.values, y_pred)

        print(
            f'{split_name} | {model_name}: '
            f'MAE={mae:.3f}, RMSE={rmse:.3f}, R2={r2:.3f}'
        )

# Compare all nine combinations
results_df = pd.DataFrame(reg_results)

print("\nAll regression results (lowest RMSE first):")
results_df.sort_values('RMSE').reset_index(drop=True)
```

```python
# 10.1 Log Transformation Experiment for BOD Regression
# The BOD target is heavily right-skewed with extreme values (>500 mg/L).
# Here, we test training models on np.log1p(y_train) to compress the target scale.
# Predictions are converted back using np.expm1() so that MAE, RMSE and R2
# are evaluated fairly on the original mg/L scale.

reg_log_results = []
reg_log_predictions = {}

for split_name, test_size in splits:

    # Split features and target (using identical splits and random state)
    X_train, X_test, y_train, y_test = train_test_split(
        X_reg, y_reg, test_size=test_size, random_state=42
    )

    # Fill missing values using training data medians only
    medians = X_train.median()
    X_train = X_train.fillna(medians)
    X_test = X_test.fillna(medians)

    # Log-transform target using np.log1p
    y_train_log = np.log1p(y_train)

    # Define the same three regression models
    reg_models = {
        'Linear Regression': LinearRegression(),
        'Decision Tree': DecisionTreeRegressor(random_state=42),
        'Random Forest': RandomForestRegressor(n_estimators=100, random_state=42)
    }

    # Train and evaluate every model
    for model_name, model in reg_models.items():
        model.fit(X_train, y_train_log)
        y_pred_log = model.predict(X_test)

        # Convert predictions back to original BOD scale using np.expm1
        y_pred = np.expm1(y_pred_log)

        mae = mean_absolute_error(y_test, y_pred)
        mse = mean_squared_error(y_test, y_pred)
        rmse = np.sqrt(mse)
        r2 = r2_score(y_test, y_pred)

        reg_log_results.append({
            'Model': model_name,
            'Split': split_name,
            'MAE': round(mae, 4),
            'MSE': round(mse, 4),
            'RMSE': round(rmse, 4),
            'R2': round(r2, 4)
        })

        # Save predictions for 80:20 visualisations
        if split_name == '80:20':
            reg_log_predictions[model_name] = (y_test.values, y_pred)

        print(
            f'{split_name} | {model_name} (Log-Target): '
            f'MAE={mae:.3f}, RMSE={rmse:.3f}, R2={r2:.3f}'
        )

# Display all log-transformed regression results
reg_log_results_df = pd.DataFrame(reg_log_results)

print("\nAll Log-Transformed Regression Results (lowest RMSE first):")
reg_log_results_df.sort_values('RMSE').reset_index(drop=True)
```

```python
# 10.5 Fair Comparison: Original Target vs Log-Transformed Target

# Merge original and log-transformed results for side-by-side comparison
comparison_df = pd.merge(
    reg_results_df,
    reg_log_results_df,
    on=['Model', 'Split'],
    suffixes=(' (Original)', ' (Log1p)')
)

display_cols = [
    'Model', 'Split',
    'MAE (Original)', 'MAE (Log1p)',
    'RMSE (Original)', 'RMSE (Log1p)',
    'R2 (Original)', 'R2 (Log1p)'
]
print('=' * 85)
print('COMPARISON: ORIGINAL TARGET vs LOG-TRANSFORMED TARGET (All on original mg/L scale)')
print('=' * 85)
print(comparison_df[display_cols].to_string(index=False))

# Bar chart comparison of MAE and RMSE for 80:20 split
comp_80 = comparison_df[comparison_df['Split'] == '80:20'].set_index('Model')

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

comp_80[['MAE (Original)', 'MAE (Log1p)']].plot(kind='bar', ax=axes[0], color=['steelblue', 'teal'], edgecolor='white', width=0.6)
axes[0].set_title('MAE Comparison (80:20 Split) - Lower is Better', fontsize=11, fontweight='bold')
axes[0].set_ylabel('MAE (mg/l)')
axes[0].tick_params(axis='x', rotation=15)

comp_80[['RMSE (Original)', 'RMSE (Log1p)']].plot(kind='bar', ax=axes[1], color=['salmon', 'darkcyan'], edgecolor='white', width=0.6)
axes[1].set_title('RMSE Comparison (80:20 Split) - Lower is Better', fontsize=11, fontweight='bold')
axes[1].set_ylabel('RMSE (mg/l)')
axes[1].tick_params(axis='x', rotation=15)

plt.suptitle('Regression Performance: Original vs Log-Transformed BOD Target', fontsize=13, fontweight='bold')
plt.tight_layout()
plt.show()

print('''
Key Observations from the Log Transformation Experiment:
1. MAE (Mean Absolute Error) improves substantially across models:
   - Random Forest MAE decreases from 4.42 to 3.84 mg/L in 80:20 split, and from 5.33 to 3.20 in 70:30 split.
   - Decision Tree MAE drops from 6.73 to 4.30 mg/L in 80:20 split.
   This proves the log transformation significantly improves prediction accuracy for the vast majority of normal river water readings.
2. R2 scores improve for tree-based models on several splits:
   - Decision Tree (80:20 split) R2 improves from -0.933 to +0.133.
   - Random Forest (70:30 split) R2 improves from -0.244 to +0.099.
   - Random Forest (65:35 split) R2 improves from -0.242 to +0.105.
3. RMSE remains high for all models:
   - Because extreme high-BOD values (>100 and >500 mg/L) are legitimate measurements preserved in the dataset, converting back with expm1() still leaves squared errors on those extreme points.
Conclusion: The log transformation is beneficial for typical BOD predictions (lower MAE), but severe right-skewness and extreme values inherently constrain raw-scale R2 when outliers are kept.
''')
```

```python
# 11.1 Create BOD categories based on CPCB-referenced thresholds

bod_col = 'B.O.D. (mg/l)'

def assign_bod_category(bod_value):
    if bod_value < 3:
        return 'Low'
    elif bod_value <= 6:
        return 'Medium'
    else:
        return 'High'

df['BOD_Category'] = df[bod_col].apply(assign_bod_category)

print('BOD Category distribution:')
print(df['BOD_Category'].value_counts())
print('\nPercentages:')
print((df['BOD_Category'].value_counts(normalize=True) * 100).round(1))
```

```python
# 11.3 Prepare classification features and target
# IMPORTANT: The original numeric BOD column is EXCLUDED to prevent target leakage!
from sklearn.metrics import precision_score, recall_score, classification_report

clf_feature_cols = [c for c in measurement_cols if c != bod_col]

X_clf = df[clf_feature_cols].copy()
y_clf = df['BOD_Category'].copy()

print('Classification features (numeric BOD excluded to prevent leakage):')
print(clf_feature_cols)
print(f'\nFeature matrix shape: {X_clf.shape}')
print(f'Target series shape:  {y_clf.shape}')
print('\nMissing values in classification features:')
print(X_clf.isnull().sum())
```

```python
# 11.4 Train classification models across splits
# Uses simple training medians to fill missing values and StandardScaler for linear/distance models

SEED = 42

def safe_split(X, y, test_size, seed):
    min_count = y.value_counts().min()
    n_required = max(2, round(1 / test_size))
    if min_count >= n_required:
        return train_test_split(X, y, test_size=test_size, random_state=seed, stratify=y)
    else:
        print(f'  Warning: Using non-stratified split (min class count={min_count})')
        return train_test_split(X, y, test_size=test_size, random_state=seed)

clf_results = []
final_clf_cm = {}  # Store (y_test, y_pred) pairs for 80:20 split

for split_name, test_size in splits:
    X_train, X_test, y_train, y_test = safe_split(X_clf, y_clf, test_size, SEED)

    # Fill missing values using training data medians only
    medians = X_train.median()
    X_train_imp = X_train.fillna(medians)
    X_test_imp  = X_test.fillna(medians)

    # Scaler fitted on train only (used for Logistic Regression and KNN)
    scaler = StandardScaler()
    X_train_sc = scaler.fit_transform(X_train_imp)
    X_test_sc  = scaler.transform(X_test_imp)

    # (model_name, model_instance, needs_scaling)
    all_clf_models = [
        ('Logistic Regression', LogisticRegression(max_iter=1000, random_state=SEED), True),
        ('Decision Tree',       DecisionTreeClassifier(random_state=SEED),             False),
        ('Random Forest',       RandomForestClassifier(n_estimators=100, random_state=SEED), False),
        ('KNN',                 KNeighborsClassifier(n_neighbors=5),                   True)
    ]

    for model_name, model, needs_scale in all_clf_models:
        X_tr = X_train_sc if needs_scale else X_train_imp
        X_te = X_test_sc  if needs_scale else X_test_imp

        model.fit(X_tr, y_train)
        y_pred = model.predict(X_te)

        acc  = accuracy_score(y_test, y_pred)
        prec = precision_score(y_test, y_pred, average='macro', zero_division=0)
        rec  = recall_score(y_test, y_pred, average='macro', zero_division=0)
        f1   = f1_score(y_test, y_pred, average='macro', zero_division=0)

        clf_results.append({
            'Model': model_name, 'Split': split_name,
            'Accuracy':          round(acc,  4),
            'Precision (macro)': round(prec, 4),
            'Recall (macro)':    round(rec,  4),
            'F1 (macro)':        round(f1,   4)
        })

        if split_name == '80:20':
            final_clf_cm[model_name] = (y_test, y_pred)

        print(f'[{split_name}] {model_name}: Acc={acc:.3f}, F1(macro)={f1:.3f}')
    print()

print('All classification models trained.')
```

```python
# 11.5 Display all classification results
clf_results_df = pd.DataFrame(clf_results)
print('Complete Classification Results (macro-averaged metrics):')
clf_results_df
```
